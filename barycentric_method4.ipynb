# Метод 4: барицентрические координаты

Проверка принадлежности точки тетраэдру ABCD, где A(1; 1; 1), B(5; 3; 1), C(2; 5; 1), D(2; 2; 5).



```python
from fractions import Fraction

# Вершины тетраэдра
A = (1, 1, 1)
B = (5, 3, 1)
C = (2, 5, 1)
D = (2, 2, 5)


def sub(p, q):
    """Вектор q -> p (разность координат)."""
    return (p[0] - q[0], p[1] - q[1], p[2] - q[2])


def det3(u, v, w):
    """Определитель 3x3 = смешанное произведение векторов u, v, w."""
    return (u[0] * (v[1] * w[2] - v[2] * w[1])
            - u[1] * (v[0] * w[2] - v[2] * w[0])
            + u[2] * (v[0] * w[1] - v[1] * w[0]))


def barycentric(P):
    """Барицентрические координаты точки P относительно ABCD."""
    AB, AC, AD, AP = sub(B, A), sub(C, A), sub(D, A), sub(P, A)
    d = det3(AB, AC, AD)
    if d == 0:
        raise ValueError("Тетраэдр вырожден")
    lb = Fraction(det3(AP, AC, AD), d)
    lc = Fraction(det3(AB, AP, AD), d)
    ld = Fraction(det3(AB, AC, AP), d)
    la = 1 - lb - lc - ld
    return la, lb, lc, ld


def classify(P):
    """Положение точки P относительно тетраэдра ABCD."""
    lambdas = barycentric(P)
    if any(l < 0 for l in lambdas):
        return "снаружи"
    zeros = sum(1 for l in lambdas if l == 0)
    return ["внутри", "на грани", "на ребре", "вершина"][zeros]


points = {
    "P1": (2, 2, 2),
    "P2": (2, 2, 1),
    "P3": (3, 2, 1),
    "P4": (1, 1, 1),
    "P5": (5, 5, 5),
}

for name, P in points.items():
    la, lb, lc, ld = barycentric(P)
    print(f"{name} = {P}: λA = {la}, λB = {lb}, λC = {lc}, λD = {ld} -> {classify(P)}")
```
